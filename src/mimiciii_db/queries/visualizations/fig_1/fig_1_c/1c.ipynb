```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import subprocess
from pathlib import Path
from IPython.display import Image, display
import warnings
warnings.filterwarnings('ignore')
import sys

from mimiciii_db import DB
from mimiciii_db.config import db_url
```

```python
db = DB.from_url(db_url())
print("Database connected successfully!")
```

```python
query = """
SELECT * FROM filtered_patients_agegrouped_with_morbidities;
"""
df = db.query_df(query)
df.head()
```

```python
# Calculate basic prevalence statistics
comorbidity_cols = [col for col in df.columns 
                   if col not in ['hadm_id', 'age', 'age_bin', 'subject_id']]
prevalence = df[comorbidity_cols].sum() / len(df) * 100
print(f"Number of comorbidities: {len(comorbidity_cols)}")
print(f"Total patients: {len(df)}")
print("\nTop 10 most prevalent comorbidities:")
print(prevalence.sort_values(ascending=False).head(10))
```
